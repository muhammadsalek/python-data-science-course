```python
# Random data
np.random.seed(42)
random_data =np.random.normal(
    loc=100,
    scale = 15,
    size = 1000
)

print("First 10 values:")
print(random_data[:10])
```

```python
# Histogram
import matplotlib.pyplot as plt # used for drawing figure(seaborn:sns)
plt.figure(figsize=(8,5))
plt.hist(random_data,bins = 30, edgecolor ="black")
plt.title("Normal Random Sample")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# Pandas
import pandas as pd
rng = np.random.default_rng(42)
n = 500 # observations
# Create Sample Dataset
df = pd.DataFrame({
    "id": np.arange(1,n+1),
    "age": rng.integers(18,80,n),
    "gender": rng.choice (
        ["Female", "Male","female","male"],
        n
    ),
    "bmi" :rng.normal(25,4,n),
    "smoker": rng.choice(
        ["Yes","No", None],
        n,
        p = [0.20, 0.75, 0.05]
    ),
    "outcome": rng.binomial(1,0.25,n)
})
```

```python
# Explore data
print(df.head())
print("\nShape:", df.shape)
print(df.dtypes)
print("\nSummary:")
print(df.describe())
```

```python
# Add missing BMI Values
missing_rows = rng.choice(
    df.index,
    25,
    replace = False
)
df.loc[missing_rows,"bmi"] = np.nan
```

```python
# Add duplicate rows
df= pd.concat(
    [df, df.iloc[:5]],
   ignore_index=True
)
print("\nAfter adding duplicates:", df.shape)
```
